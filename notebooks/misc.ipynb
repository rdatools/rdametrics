```python
from collections import defaultdict

by_combo = dict()
for variant, _data in by_variant.items():
    for combo, _measures in _data.items():
        if combo not in by_combo:
            by_combo[combo] = defaultdict(list)
        for measure, value in _measures.items():
            by_combo[combo][measure].append(value)

by_combo_avg = dict()
for combo, measures in by_combo.items():
    by_combo_avg[combo] = dict()
    for measure, values in measures.items():
        by_combo_avg[combo][measure] = sum(values) / len(values)

print("{")
for combo, _measures in by_combo_avg.items():
    print(f'  "{combo}": {{')
    for m, value in _measures.items():
        formatted_value: str = f'{value:.10f}'
        print(f'      "{m}": {formatted_value},')
    print("    },")
print("}")
```
